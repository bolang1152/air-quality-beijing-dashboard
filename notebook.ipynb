### Cleaning Data

Pada tahap ini, dilakukan proses pembersihan data berdasarkan permasalahan yang ditemukan pada tahap Assessing Data. Proses pembersihan difokuskan pada penanganan missing values agar dataset memiliki kualitas yang lebih baik dan dapat digunakan untuk analisis selanjutnya. Setelah proses pembersihan selesai, dilakukan pemeriksaan kembali untuk memastikan data sudah siap digunakan.

#### Fixing Missing Values

```python
df_clean = df.copy()

df_clean = df_clean.drop_duplicates()

df_clean = df_clean.sort_values(
    by=["station", "year", "month", "day", "hour"]
).reset_index(drop=True)

numeric_cols = [
    "PM2.5", "PM10", "SO2", "NO2", "CO", "O3",
    "TEMP", "PRES", "DEWP", "RAIN", "WSPM"
]

df_clean[numeric_cols] = (
    df_clean.groupby("station")[numeric_cols]
    .transform(
        lambda x: x.interpolate(
            method="linear",
            limit_direction="both"
        )
    )
)

df_clean["wd"] = (
    df_clean.groupby("station")["wd"]
    .transform(lambda x: x.fillna(x.mode().iloc[0]))
)

df_clean.info()
```

##### Mengidentifikasi outlier menggunakan metode IQR

```python
outlier_cols = [
    "PM2.5", "PM10", "SO2", "NO2", "CO", "O3",
    "TEMP", "PRES", "DEWP", "RAIN", "WSPM"
]

outlier_results = []

for col in outlier_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (df_clean[col] < lower_bound) |
        (df_clean[col] > upper_bound)
    ).sum()

    outlier_results.append({
        "Column": col,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count
    })

outlier_df = pd.DataFrame(outlier_results)
outlier_df
```

##### Memeriksa missing values, data duplikat dan ukuran dataset

```python
print("Missing Values:")
print(df_clean.isna().sum())

print("\nJumlah Data Duplikat:")
print(df_clean.duplicated().sum())

print("\nUkuran Dataset:")
print(df_clean.shape)
```

### 4. Explore Konsentrasi PM2.5 Berdasarkan Musim
Pada tahap ini, setiap data dikelompokkan ke dalam empat musim: **musim semi** (Maret-Mei), **musim panas** (Juni-Agustus), **musim gugur** (September-November), dan **musim dingin** (Desember-Februari). Kemudian dihitung rata-rata PM2.5 pada setiap musim.

```python
def get_season(month):
    if month in [3, 4, 5]:
        return "Musim Semi"
    elif month in [6, 7, 8]:
        return "Musim Panas"
    elif month in [9, 10, 11]:
        return "Musim Gugur"
    return "Musim Dingin"

df_clean["season"] = df_clean["month"].apply(get_season)

season_order = ["Musim Semi", "Musim Panas", "Musim Gugur", "Musim Dingin"]

pm25_by_season = (
    df_clean.groupby("season")["PM2.5"]
    .mean()
    .round(2)
    .reindex(season_order)
)

print("Musim dengan rata-rata PM2.5 tertinggi:",
      pm25_by_season.idxmax(), "=", pm25_by_season.max())
print("Musim dengan rata-rata PM2.5 terendah:",
      pm25_by_season.idxmin(), "=", pm25_by_season.min())

pm25_by_season
```

**Insight:**

Rata-rata PM2.5 untuk keempat musim ditampilkan pada keluaran di atas. Pengelompokan ini memungkinkan perbandingan langsung konsentrasi PM2.5 antarmusim, termasuk musim dengan rata-rata tertinggi dan terendah, sesuai Pertanyaan 3.

Pada variabel meteorologi, suhu (TEMP) memiliki hubungan positif yang kuat dengan titik embun (DEWP) sebesar 0,82. Sebaliknya, suhu memiliki hubungan negatif yang kuat dengan tekanan udara (PRES) sebesar -0,81.

Kecepatan angin (WSPM) memiliki hubungan negatif dengan NO2 sebesar -0,40 dan PM2.5 sebesar -0,27. Hal ini menunjukkan bahwa peningkatan kecepatan angin cenderung berkaitan dengan penurunan konsentrasi kedua polutan tersebut.

Sementara itu, curah hujan (RAIN) memiliki hubungan yang sangat lemah dengan sebagian besar polutan. Secara keseluruhan, hasil korelasi menunjukkan bahwa beberapa polutan memiliki hubungan yang cukup kuat satu sama lain dan kondisi meteorologi juga memiliki keterkaitan dengan konsentrasi polutan. Namun, korelasi tidak selalu menunjukkan hubungan sebab-akibat.

## Analisis Lanjutan
### Analisis Pengaruh Kondisi Meteorologi terhadap Konsentrasi PM2.5
Pada analisis lanjutan ini, dilakukan pengamatan terhadap hubungan antara kondisi meteorologi dengan konsentrasi PM2.5. Variabel yang digunakan meliputi suhu udara (TEMP), tekanan udara (PRES), titik embun (DEWP), curah hujan (RAIN), dan kecepatan angin (WSPM).

Analisis ini bertujuan untuk mengetahui bagaimana perubahan kondisi meteorologi berkaitan dengan konsentrasi PM2.5. Pengamatan dilakukan dengan membandingkan rata-rata konsentrasi PM2.5 pada beberapa kelompok kondisi meteorologi.

#### Membuat kategori kecepatan angin

```python
df_analysis = df_clean.copy()

df_analysis["Wind_Category"] = pd.cut(
    df_analysis["WSPM"],
    bins=[-0.01, 1, 2, 3, float("inf")],
    labels=["Rendah", "Sedang", "Tinggi", "Sangat Tinggi"]
)

pm25_by_wind = (
    df_analysis.groupby("Wind_Category", observed=False)["PM2.5"]
    .mean()
    .round(2)
)

pm25_by_wind
```

#### Visualisasi hubungan kecepatan angin dan PM2.5

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    x=pm25_by_wind.index,
    y=pm25_by_wind.values
)

plt.title("Rata-rata PM2.5 Berdasarkan Kategori Kecepatan Angin")
plt.xlabel("Kategori Kecepatan Angin")
plt.ylabel("Rata-rata PM2.5")
plt.tight_layout()
plt.show()
```

#### Melihat hubungan suhu dan PM2.5

```python
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df_analysis.sample(5000, random_state=42),
    x="TEMP",
    y="PM2.5",
    alpha=0.4
)

plt.title("Hubungan Suhu Udara dengan Konsentrasi PM2.5")
plt.xlabel("Suhu Udara (°C)")
plt.ylabel("Konsentrasi PM2.5")
plt.tight_layout()
plt.show()
```

**Insight:**

Berdasarkan hasil analisis lanjutan, terdapat pola hubungan antara kondisi meteorologi dengan konsentrasi PM2.5.

1. **Kecepatan Angin terhadap PM2.5**

   Hasil pengelompokan menunjukkan bahwa semakin tinggi kecepatan angin, rata-rata konsentrasi PM2.5 cenderung semakin rendah. Pada kategori angin rendah, rata-rata PM2.5 mencapai 101,90. Nilai tersebut menurun menjadi 84,64 pada kategori sedang, 59,39 pada kategori tinggi, dan 35,45 pada kategori sangat tinggi.

   Pola ini menunjukkan bahwa kecepatan angin memiliki hubungan berbanding terbalik dengan konsentrasi PM2.5. Angin yang lebih kencang dapat membantu menyebarkan polutan sehingga konsentrasinya di suatu area cenderung berkurang.

2. **Suhu Udara terhadap PM2.5**

   Berdasarkan scatter plot, lonjakan konsentrasi PM2.5 yang tinggi lebih banyak ditemukan pada rentang suhu rendah, khususnya sekitar -10°C hingga 5°C. Sementara itu, pada suhu di atas 20°C, nilai konsentrasi PM2.5 yang sangat tinggi terlihat lebih jarang.

   Meskipun sebagian besar data pada berbagai rentang suhu terkonsentrasi pada nilai PM2.5 di bawah 150, pola tersebut menunjukkan bahwa kondisi suhu rendah berkaitan dengan kemungkinan munculnya konsentrasi PM2.5 yang lebih tinggi.

**Kesimpulan Analisis Lanjutan:**

Kecepatan angin dan suhu udara menunjukkan pola hubungan yang berbeda terhadap konsentrasi PM2.5. Kecepatan angin memiliki pola penurunan rata-rata PM2.5 yang konsisten pada setiap kategori, sedangkan suhu udara memperlihatkan bahwa lonjakan konsentrasi PM2.5 yang tinggi lebih sering ditemukan pada kondisi suhu rendah. Hasil ini memperkuat pemahaman bahwa kondisi meteorologi berkaitan dengan perubahan kualitas udara di Beijing.
